```python
import random

# Define responses
intro_responses = ["Hi there! I see you're interested in the heart failure dataset. What would you like to know?",
                   "Hello! How can I help you with the heart failure dataset?",
                   "Hi! Ready to dive into the heart failure dataset? What questions do you have?"]
goodbye_responses = ["Goodbye! If you have more questions later, feel free to come back.",
                     "Alright, take care! Don't hesitate to return if you need further assistance.",
                     "Farewell! Feel free to reach out again if you need more information."]
help_responses = ["Sure, I'm here to assist you with any questions you have about the heart failure dataset. What do you need help with?",
                  "Of course! Feel free to ask me anything about the heart failure dataset.",
                  "Absolutely! I'm here to provide support and answer your questions about the heart failure dataset."]

# Function to respond to user input
def respond_to_user_input(user_input):
    user_input = user_input.lower()
    if 'hello' in user_input or 'hi' in user_input or 'hey' in user_input:
        return random.choice(intro_responses)
    elif 'goodbye' in user_input or 'bye' in user_input:
        return random.choice(goodbye_responses)
    elif 'help' in user_input:
        return random.choice(help_responses)
    elif 'heart failure' in user_input:
        return "The heart failure dataset contains medical records of patients, including various clinical features such as age, anaemia, diabetes, ejection fraction, etc. What specific information are you looking for?"
    elif 'age' in user_input:
        return "Age represents the age of the patient in years. What else would you like to know?"
    elif 'anaemia' in user_input:
        return "Anaemia indicates a decrease in red blood cells or haemoglobin. Do you need more details?"
    elif 'creatinine phosphokinase' in user_input or 'cpk' in user_input:
        return "CPK is the level of the CPK enzyme in the blood, measured in mcg/L. What else can I help you with?"
    elif 'diabetes' in user_input:
        return "Diabetes indicates whether the patient has diabetes or not. What other information do you need?"
    elif 'ejection fraction' in user_input:
        return "Ejection fraction is the percentage of blood leaving the heart at each contraction. What else would you like to know?"
    elif 'high blood pressure' in user_input:
        return "High blood pressure indicates whether the patient has hypertension or not. Do you have any other questions?"
    elif 'platelets' in user_input:
        return "Platelets represent the platelet count in the blood, measured in kiloplatelets/mL. What specific details do you need?"
    elif 'sex' in user_input:
        return "Sex indicates whether the patient is male or female. What other information can I provide?"
    elif 'serum creatinine' in user_input:
        return "Serum creatinine is the level of serum creatinine in the blood, measured in mg/dL. Anything else you'd like to know?"
    elif 'serum sodium' in user_input:
        return "Serum sodium represents the level of serum sodium in the blood, measured in mEq/L. How else can I assist you?"
    elif 'smoking' in user_input:
        return "Smoking indicates whether the patient smokes or not. Do you need more details?"
    elif 'time' in user_input:
        return "Time represents the follow-up period in days. What other questions do you have?"
    elif 'death event' in user_input:
        return "DEATH_EVENT indicates whether the patient died during the follow-up period. Is there anything else you'd like to know?"
    else:
        return "I'm sorry, I may not be able to assist with that specific question. Please ask about a feature from the heart failure dataset, and I'll do my best to provide information."

# Main loop to interact with the user
def main():
    print("Welcome to Heart Failure Dataset Support Chatbot!")
    print("How can I assist you today? (Type 'exit' to end the conversation)")

    while True:
        user_input = input("You: ")
        if user_input.lower() == 'exit':
            print(random.choice(goodbye_responses))
            break
        response = respond_to_user_input(user_input)
        print("Bot:", response)

if __name__ == "__main__":
    main()
```
